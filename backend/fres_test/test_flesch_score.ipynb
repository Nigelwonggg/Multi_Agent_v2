## Interactive Input: Add Your Own Multi-Line Text Samples

Use this section to input your own text samples and automatically save them to a JSONL file.

```python
def add_sample_to_jsonl(agent_text, gpt_text, gemini_text, output_file='my_samples.jsonl'):
    """
    Add a new sample to a JSONL file.
    Creates the file if it doesn't exist, appends if it does.
    
    Parameters:
    - agent_text: Multi-line text from your agent/framework
    - gpt_text: Multi-line text from GPT baseline
    - gemini_text: Multi-line text from Gemini baseline
    - output_file: Path to the JSONL file (default: 'my_samples.jsonl')
    """
    sample = {
        "agent": agent_text.strip(),
        "gpt": gpt_text.strip(),
        "gemini": gemini_text.strip()
    }
    
    # Append to file (creates if doesn't exist)
    with open(output_file, 'a', encoding='utf-8') as f:
        f.write(json.dumps(sample, ensure_ascii=False) + '\n')
    
    return sample

# Helper to view all samples in a file
def view_samples_in_jsonl(filename='my_samples.jsonl'):
    """View all samples currently in the JSONL file"""
    if not os.path.exists(filename):
        print(f"File '{filename}' does not exist yet. Add samples to create it.")
        return []
    
    samples = []
    with open(filename, 'r', encoding='utf-8') as f:
        for i, line in enumerate(f, 1):
            sample = json.loads(line)
            samples.append(sample)
    
    print(f"Found {len(samples)} sample(s) in {filename}")
    return samples

print("✓ Helper functions loaded")
print("  - add_sample_to_jsonl(): Add a new sample")
print("  - view_samples_in_jsonl(): View existing samples")
```
